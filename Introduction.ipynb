# Introduction to Consumers Behavior (2026) Dataset

```python
import kagglehub

# Download latest version
path = kagglehub.dataset_download("sohaibdevv/consumer-shopping-behavior-and-preference-study-2026")

print("Path to dataset files:", path)
```

```python
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
file_list = os.listdir(path)
print("Dataset:", file_list)

# Print first 5 rows of dataset
df = pd.read_csv(path + "/" + file_list[0])
df.head()
```

```python
# Print information of Dataset
df.info()
```

```python
# Display the descriptive statistics for numerical columns
display(df.describe())
```

```python
behavioral_scores = ['tech_savvy_score', 'brand_loyalty_score', 'impulse_buying_score', 'environmental_awareness']

# Group by Gender and calculate the mean of the specified behavioral scores
gender_behavior_analysis = df.groupby('gender')[behavioral_scores].mean()

print("Analysis of Behavioral Scores by Gender:")
print(gender_behavior_analysis)
```

# Explanation:
 The analysis of behavioral scores by gender revelas:
 - **tech_savvy_score**: Females have a slightly higher average tech-savvy score compared to Males and Others.
 - **brand_loyalty_score**: All gender categories show similar averag brand loyalty scores that indicate little differences.
 - **impulse_buying_score**: Females exhibit a slightly higher average impulse buying score.
 - **environmental_awareness**: Females show a marginally higher average environmental awareness score.

 Overall, there are minor differences that exist, but the behavioral scores are quite similar across the gender categories, which might suggest that gender is not a primary driver of these specific behavioral traits in the dataset.

```python
city_tier_analysis = df.groupby('city_tier')[['tech_savvy_score', 'brand_loyalty_score', 'impulse_buying_score', 'environmental_awareness']].mean()

print("Analysis of Behavioral Scores by City Tier:")
print(city_tier_analysis)
```

The table displays the average behavioral scores (*tech_savvy_score*, *brand_loyalty_score*, *impulse_buying_score*, *environmental_awareness*) across different city tiers:
- **Minimal Variation**: For all behavioral scores, the average values are close across Tiers 1, 2, and 3 cities,
- **No Distinct Trends**: There aren't any trends or significant differences that indicate that residents of one city tier are notably more or less *tech_savvy*, *brand_loyalty*, *impulsive*, or *environmentally_aware* than those in other tiers.

It may suggest that *city_tier* doesnt have significant influence on these particular behavioral scores, that could imply that these traits are moreuniversal or influenced by factors other than urban classification.

```python
# Group by age group and calculate the mean for behavioral scores
age_behavior_variation = df.groupby('age_group', observed=False)[behavioral_scores].mean().reset_index()

print("Average behavioral scores by Age Group:")
display(age_behavior_variation)

# Visualize the variation by age group
fig, axes = plt.subplots(2, 2, figsize=(16, 12), sharex=True)
axes = axes.flatten()

for i, score in enumerate(behavioral_scores):
    sns.barplot(x='age_group', y=score, data=age_behavior_variation, ax=axes[i], palette='viridis')
    axes[i].set_title(f'Average {score.replace("_", " ").title()} by Age Group')
    axes[i].set_xlabel('Age Group')
    axes[i].set_ylabel(score.replace("_", " ").title())
    axes[i].tick_params(axis='x', rotation=45)

plt.tight_layout()
plt.show()
```

# Explanation:

The table illustrates various behavioral scores that change across different age groups:
- **tech_savvy_score**: Peaks in the younger age groups (18-24, 25-34) and steadily declines with increasing age. It aligns with general expectations of digital native generations.
- **brand_loyalty_score**: Shows a more complex pattern, generally increasing from youger groups, peaking in middle-aged groups (35-44, 45-54), and then slightly decreasing in older groups. This suggests that brand loyalty may develop and solidify over time.
- **impulse_buying_score**: Tends to be higher in younger age groups (18-24, 25-34) and gradually decreases as consumers get older. Younger consumers might be more prone to spontaneous purchases.
- **environmental_awareness**: Similar to tech-saviness and impulse buying, environmental awareness scores are generally higher in younger age groups and show a slight decline with increasing age

*age_group* appears to be a notable factor that influences behavioral scores, with the younger demographic often scoring higher in tech-saviness, impulsivity, and environmental awareness, while brand loyalty tends to be the strongest in middle-aged consumers,
