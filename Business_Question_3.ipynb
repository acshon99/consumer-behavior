# Question 3:

Is there a relationship between digital engagement metrics (*daily_internet_hours*, *social_media_hours*) and a consumer's *impulse_buying_score* or *online_payment_trust_score*?

```python
import os
import kagglehub
import pandas as pd

# Download latest version to ensure 'path' is defined
path = kagglehub.dataset_download("sohaibdevv/consumer-shopping-behavior-and-preference-study-2026")
print("Path to dataset files:", path)

# List the contents of the downloaded directory to find the correct CSV file
print("Contents of the dataset directory:")
for root, dirs, files in os.walk(path):
    for name in files:
        print(os.path.join(root, name))
    for name in dirs:
        print(os.path.join(root, name))

# Construct the full path to the CSV file with the correct name
csv_file_path = os.path.join(path, 'Consumer_Shopping_Trends_2026 (6).csv')

# Load the dataset into a pandas DataFrame
df = pd.read_csv(csv_file_path)

print(f"DataFrame 'df' loaded with shape: {df.shape}")
print(df.head())
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Define the columns of interest
digital_engagement_metrics = ['daily_internet_hours', 'social_media_hours']
consumer_behavior_metrics = ['impulse_buying_score', 'online_payment_trust_score']

# Calculate the correlation matrix for the relevant columns
correlation_matrix = df[digital_engagement_metrics + consumer_behavior_metrics].corr()

print("Correlation Matrix:")
print(correlation_matrix)

# Visualize the correlations using a heatmap
plt.figure(figsize=(8, 6))
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', fmt=".2f")
plt.title('Correlation between Digital Engagement and Consumer Behavior Metrics')
plt.show()
```

```python
# Correlation between daily_internet_hours and online_payment_trust_score
corr_internet_trust = df['daily_internet_hours'].corr(df['online_payment_trust_score'])
print(f"Correlation between daily_internet_hours and online_payment_trust_score: {corr_internet_trust}")

# Correlation between social_media_hours and online_payment_trust_score
correlation_matrixt_social = df['social_media_hours'].corr(df['online_payment_trust_score'])
print(f"Correlation between social_media_hours and online_payment_trust_score: {correlation_matrixt_social}")

# Correlation between social_media_hours and online_payment_trust_score
correlation_matrix_internet = df['online_payment_trust_score'].corr(df['social_media_hours'])
print(f"Correlation between daily_internet_hours and social_media_hours: {correlation_matrix_internet}")
```

# Explanation:

The correlation analysis is usd to understand the linear relationships between digital engagement metricd and specific buying behaviors.

- **Daily Internet Hours vs. Online Payment Trust Score**: The Peardon correlation coefficient is 0.0147, which is close to zero. It indicates a negligible linear relationship between the amount of time spent online and a consumer's trust in online payments.
- **Social Media Hours vs. Impulse Buying Score**: The Pearson correlation coefficient is 0.0116, also close to zero. This suggests that a negligible linear relationship between social media usage and a consumer's tendency for impulse buying.
- **Social Media Hours vs. Online Payment Trust Score**: The correlation coefficient is 0.0027, which is almost zero. It implies that there is virtually no linear relationship between social media hours and online payment trust.

In conclusion, based on these linear correlation coefficients, there is no significant linear relationship observed between digital engagement metrics (*daily_internet_hours*, *social_media_hours*) and a consumer's *impulse_buying_score* or *online_payment_trust_score* in this dataset.
