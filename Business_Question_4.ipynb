# Question 4:

How do *discount_sensitivity*, *return_frequency*, *delivery_fee_sensitivity*, and *free_returm_importance* vary across different *shopping_preference* groups or *city_tier*?

```python
import os
import kagglehub
import pandas as pd

# Download latest version to ensure 'path' is defined
path = kagglehub.dataset_download("sohaibdevv/consumer-shopping-behavior-and-preference-study-2026")
print("Path to dataset files:", path)

# List the contents of the downloaded directory to find the correct CSV file
print("Contents of the dataset directory:")
for root, dirs, files in os.walk(path):
    for name in files:
        print(os.path.join(root, name))
    for name in dirs:
        print(os.path.join(root, name))

# Construct the full path to the CSV file with the correct name
csv_file_path = os.path.join(path, 'Consumer_Shopping_Trends_2026 (6).csv')

# Load the dataset into a pandas DataFrame
df = pd.read_csv(csv_file_path)

print(f"DataFrame 'df' loaded with shape: {df.shape}")
print(df.head())
```

```python
# Variation by  Shopping Preference Category
preference_factors_variation = df.groupby('shopping_preference').agg({
    'discount_sensitivity': 'mean',
    'return_frequency': 'mean',
    'delivery_fee_sensitivity': 'mean',
    'free_return_importance': 'mean'
})

print("Variation by Shopping Preference Category:")
print(preference_factors_variation)
```

# Explanation:

The table above shows the average vaues of *discount_sensitivity*, *return_frequency*, *delivery_fee_sensitivity*, and *free_return_importance* across the different *shopping_preference* categories:
- **Discount Sensivity**: Hybrid shoppers show slightly higher discount sensitivity compared to Online and Store shoppers, that have similar average scores.
- **Return frequency**: Online shoppers have a slightly higher *return_frequency* on average, which could be epected given the nature of online purchases where physical inspection is not possible before buying.
- **Delivery Fee Sensitivity**: Online shoppers are noticeably more sensitive to delivery fees, which makes sense as delivery is a direct cost associated with their preferred shoppng method. Store shoppers are least sensitive.
- **Free Return Importance**: Online shoppers also place higher importance on free returns, aligning with the typical challenges and expectations of online shopping.

These findings suggest that there are some overlaps, and that each shopping preference group has a distinct concerns regarding transactional factors, particularly with Online shoppers being more sensitive to delivery costs and returm policies.
